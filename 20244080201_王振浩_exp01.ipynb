```python
# 实验2：成绩等级判断函数

def score_level(score):
    if score < 0 or score > 100:
        raise ValueError("分数必须在0到100之间")

    if score >= 90:
        return "优秀"
    elif score >= 80:
        return "良好"
    elif score >= 60:
        return "及格"
    else:
        return "不及格"


# 测试不同分数
print("95分：", score_level(95))
print("85分：", score_level(85))
print("70分：", score_level(70))
print("50分：", score_level(50))

# 测试超出范围的分数
try:
    print(score_level(101))
except ValueError as e:
    print("出现异常：", e)
```

**成绩等级判断函数说明：**

1. 使用 `def` 定义 `score_level(score)` 函数。
2. 使用 `if` 和 `elif` 判断分数所在的范围，并返回对应的成绩等级。
3. 当分数大于等于90分时返回“优秀”，大于等于80分时返回“良好”，大于等于60分时返回“及格”，其余有效分数返回“不及格”。
4. 当分数小于0或大于100时，使用 `raise ValueError` 抛出异常，提示分数超出范围。
5. 使用 `try-except` 捕获异常，避免程序因为输入了超出范围的分数而直接中断。

```python
# 实验2：使用 for 和 while 计算 1 到 100 的整数和

# 方法一：使用 for 循环
sum_for = 0
for i in range(1, 101):
    sum_for += i

# 方法二：使用 while 循环
sum_while = 0
i = 1
while i <= 100:
    sum_while += i
    i += 1

# 输出结果
print("for循环计算结果：", sum_for)
print("while循环计算结果：", sum_while)
print("两种方法结果是否相同：", sum_for == sum_while)
```

**循环求和说明：**

1. `for` 循环通过 `range(1, 101)` 依次取出1到100的整数，并累加到 `sum_for` 中。
2. `while` 循环在 `i <= 100` 的条件成立时持续累加，每次循环后将 `i` 加1。
3. 两种方法的计算结果都是5050，说明两种循环都能完成1到100的整数求和。`for` 循环适合遍历确定范围的数据，`while` 循环则根据条件决定是否继续执行。

```python
# 实验2：递归和迭代计算阶乘

# 方法一：递归计算阶乘
def factorial_recursive(n):
    if n == 0:
        return 1
    return n * factorial_recursive(n - 1)


# 方法二：迭代计算阶乘
def factorial_iterative(n):
    result = 1
    for i in range(1, n + 1):
        result *= i
    return result


# 比较不同 n 的计算结果
for n in [0, 1, 5, 10]:
    recursive_result = factorial_recursive(n)
    iterative_result = factorial_iterative(n)

    print(f"n = {n}")
    print(f"递归结果：{recursive_result}")
    print(f"迭代结果：{iterative_result}")
    print(f"结果是否相同：{recursive_result == iterative_result}")
    print()
```

**阶乘计算说明：**

1. 递归方法通过函数调用自身计算阶乘。当 `n == 0` 时返回1，否则返回 `n * factorial_recursive(n - 1)`。
2. 迭代方法使用 `for` 循环，从1开始依次相乘，得到阶乘结果。
3. 根据运行结果，n为0、1、5、10时，两种方法的计算结果分别为1、1、120、3628800，结果均相同。
4. 递归方法通过函数调用自身实现计算，迭代方法则通过循环重复执行计算，两种方法都可以求出阶乘。

```python
# 实验2：统计列表的最小值、最大值和平均值

def summary(values):
    minimum = min(values)
    maximum = max(values)
    average = sum(values) / len(values)
    return minimum, maximum, average


# 创建一个数值列表
scores = [85, 58, 92, 76, 60, 89]

# 使用元组解包接收返回结果
minimum, maximum, average = summary(scores)

# 输出统计结果
print("最小值：", minimum)
print("最大值：", maximum)
print("平均值：", average)
```

**统计函数说明：**

1. 使用 `def` 定义 `summary(values)` 函数，接收一个数值列表。
2. 使用 `min()` 和 `max()` 分别计算列表中的最小值和最大值。
3. 使用 `sum(values) / len(values)` 计算平均值。
4. 使用 `return` 将最小值、最大值和平均值作为元组返回。
5. 调用函数时，使用 `minimum, maximum, average = summary(scores)` 接收返回的三个结果，这种方式称为元组解包。

```python
# 实验3：使用 zip() 创建姓名与成绩的字典

names = ["Alice", "Bob", "Carol", "David", "Eva", "Frank"]
scores = [85, 58, 92, 76, 60, 89]

# 使用 zip() 创建字典
student_scores = dict(zip(names, scores))

# 使用 items() 逐项输出姓名和成绩
for name, score in student_scores.items():
    print(name, "的成绩是：", score)
```

**使用 zip() 创建字典的说明：**

1. 使用 `zip(names, scores)` 将姓名列表和成绩列表中的元素一一配对。
2. 使用 `dict()` 将配对结果转换为字典，其中姓名作为键，成绩作为值。
3. 使用 `items()` 获取字典中的键值对，再通过 `for` 循环逐项输出每位学生的姓名和成绩。

```python
# 实验3：使用 enumerate() 输出学生信息

for index, (name, score) in enumerate(zip(names, scores), start=1):
    print(f"序号：{index}，姓名：{name}，成绩：{score}")
```

**使用 enumerate() 输出学生信息的说明：**

1. 使用 `zip(names, scores)` 将姓名和成绩一一配对。
2. 使用 `enumerate()` 遍历配对后的学生信息，并通过 `start=1` 设置序号从1开始。
3. 使用 `for` 循环依次获取学生序号、姓名和成绩，再使用字符串格式化输出。

```python
# 实验3：使用列表推导式处理成绩

# 提取及格学生的姓名
passed_names = [name for name, score in zip(names, scores) if score >= 60]

# 生成所有成绩的平方列表
squared_scores = [score ** 2 for score in scores]

# 输出结果
print("及格学生：", passed_names)
print("所有成绩的平方：", squared_scores)
```

**列表推导式说明：**

1. 使用列表推导式和 `zip(names, scores)` 将姓名与成绩配对，通过 `if score >= 60` 筛选出及格学生的姓名。
2. 使用列表推导式 `[score ** 2 for score in scores]` 计算所有成绩的平方，并将结果保存到新列表中。
3. 根据运行结果，及格学生有 Alice、Carol、David、Eva 和 Frank；所有成绩的平方分别为7225、3364、8464、5776、3600和7921。

```python
# 实验3：成绩统计

# 计算人数、平均分、最高分和最低分
student_count = len(scores)
average_score = sum(scores) / len(scores)
highest_score = max(scores)
lowest_score = min(scores)

# 计算及格人数和及格率
pass_count = sum(1 for score in scores if score >= 60)
pass_rate = pass_count / student_count

# 输出统计结果
print("总人数：", student_count)
print(f"平均分：{average_score:.2f}")
print("最高分：", highest_score)
print("最低分：", lowest_score)
print(f"及格率：{pass_rate:.2%}")
```

**成绩统计说明：**

1. 使用 `len(scores)` 计算学生总人数，使用 `sum(scores) / len(scores)` 计算平均分。
2. 使用 `max(scores)` 和 `min(scores)` 分别计算最高分和最低分。
3. 使用列表中的成绩判断及格人数，分数大于等于60分视为及格，再用及格人数除以总人数得到及格率。
4. 使用 `:.2f` 将平均分保留两位小数，使用 `:.2%` 将及格率转换为百分数并保留两位小数。
5. 统计结果为：总人数6人，平均分76.67分，最高分92分，最低分58分，及格率83.33%。

3. 使用 `with open(file_path, "w", encoding="utf-8")` 打开文件，并将学生成绩数据写入 `students.txt`。其中，`w` 表示写入模式，`encoding="utf-8"` 表示使用UTF-8编码。
4. 使用 `with open(file_path, "r", encoding="utf-8")` 读取文件，再通过 `f.read()` 获取文件中的全部内容。
5. 文件中包含正常成绩和异常数据，为后续练习数据解析与异常处理做好准备。

```python
# 实验4：编写学生数据解析函数

def parse_student(line):
    # 按逗号拆分数据
    fields = line.strip().split(",")

    # 检查字段数量
    if len(fields) != 2:
        raise ValueError("每条记录必须包含姓名和成绩两个字段")

    # 检查姓名是否为空
    name = fields[0].strip()
    if not name:
        raise ValueError("姓名不能为空")

    # 将成绩转换为整数
    try:
        score = int(fields[1].strip())
    except ValueError:
        raise ValueError("成绩必须是整数")

    # 检查成绩范围
    if score < 0 or score > 100:
        raise ValueError("成绩必须在0到100之间")

    # 返回有效的学生记录
    return {"name": name, "score": score}


# 测试函数
print(parse_student("Alice,85"))
```

```python
# 实验4：逐行读取文件并处理无效记录

valid_records = []
invalid_records = []

with open(file_path, "r", encoding="utf-8") as f:
    # 跳过第一行表头
    next(f)

    # 从第2行开始处理数据
    for line_number, line in enumerate(f, start=2):
        try:
            student = parse_student(line)
            valid_records.append(student)
        except ValueError as e:
            invalid_records.append({
                "line": line_number,
                "content": line.strip(),
                "error": str(e)
            })

# 输出有效记录
print("有效记录：")
for student in valid_records:
    print(student)

# 输出无效记录及错误原因
print("\n无效记录：")
for record in invalid_records:
    print(
        f"第{record['line']}行，"
        f"内容：{record['content']}，"
        f"错误原因：{record['error']}"
    )

# 输出记录数量
print("\n有效记录数：", len(valid_records))
print("无效记录数：", len(invalid_records))
```

**数据解析与异常处理说明：**

1. 编写 `parse_student(line)` 函数，使用 `split(",")` 将每行数据拆分为姓名和成绩。
2. 检查每条记录的字段数量，判断姓名是否为空，并将成绩转换为整数。
3. 如果字段数量不正确、姓名为空、成绩不能转换为整数或成绩不在0到100之间，则使用 `raise ValueError` 抛出异常。
4. 使用 `try-except` 逐行处理文件中的数据。如果记录有效，就将其保存到字典组成的列表中；如果记录无效，就记录行号、原始内容和错误原因，并继续处理后面的记录。
5. 本次实验共读取到6条有效记录和3条无效记录。无效记录分别是 Grace 的成绩不是整数、Henry 的成绩超出范围，以及 Incomplete 的字段数量不正确。

```python
# 实验4：统计有效和无效记录，并写入 report.txt

# 1. 统计有效记录数和无效记录数
valid_count = len(valid_records)
invalid_count = len(invalid_records)

# 2. 提取有效记录中的成绩
scores_valid = [record["score"] for record in valid_records]

# 3. 计算平均分
average_score = sum(scores_valid) / len(scores_valid)

# 4. 计算及格率，成绩达到60分视为及格
pass_count = sum(1 for score in scores_valid if score >= 60)
pass_rate = pass_count / len(scores_valid)

# 5. 构造统计报告的内容
report_content = (
    f"有效记录数：{valid_count}\n"
    f"无效记录数：{invalid_count}\n"
    f"平均分：{average_score:.2f}\n"
    f"及格率：{pass_rate:.2%}\n"
)

# 6. 使用 w 模式写入 report.txt
report_path = os.path.join(data_dir, "report.txt")

with open(report_path, "w", encoding="utf-8") as f:
    f.write(report_content)

# 7. 读取并查看报告内容
print("report.txt 的内容：")
with open(report_path, "r", encoding="utf-8") as f:
    print(f.read())

# 8. 单独演示 w 和 a 模式
demo_path = os.path.join(data_dir, "mode_demo.txt")

# 使用 w 模式写入第一行
with open(demo_path, "w", encoding="utf-8") as f:
    f.write("第一行：使用 w 模式写入。\n")

# 使用 a 模式追加第二行
with open(demo_path, "a", encoding="utf-8") as f:
    f.write("第二行：使用 a 模式追加。\n")

# 读取文件，查看两种模式的效果
print("文件模式演示：")
with open(demo_path, "r", encoding="utf-8") as f:
    print(f.read())
```

**统计结果与文件打开模式说明：**

1. 使用 `len(valid_records)` 和 `len(invalid_records)` 分别统计有效记录数和无效记录数。
2. 使用有效记录中的成绩计算平均分，并将成绩大于等于60分的记录视为及格记录，计算及格率。
3. 使用 `open(report_path, "w", encoding="utf-8")` 将有效记录数、无效记录数、平均分和及格率写入 `report.txt`。
4. `w` 模式用于写入文件。如果文件已经存在，原有内容会被覆盖；如果文件不存在，则会创建新文件。
5. `a` 模式用于追加内容。如果文件已经存在，新内容会添加到原有内容后面，不会覆盖原有内容；如果文件不存在，则会创建新文件。
6. 本次实验统计得到6条有效记录、3条无效记录、平均分76.67分、及格率83.33%。通过文件模式演示，可以观察到 `w` 模式用于写入，而 `a` 模式用于追加内容。

```python
# 实验5：类、继承与对象

# 1. 定义 Student 类
class Student:
    def __init__(self, name, score):
        self.name = name
        self.score = score

    # 判断学生是否及格
    def is_passed(self):
        return self.score >= 60

    # 定义对象的字符串表示
    def __str__(self):
        return f"{self.name}（成绩：{self.score}）"


# 2. 根据实验4中的有效记录创建 Student 对象
students = []

for record in valid_records:
    student = Student(record["name"], record["score"])
    students.append(student)

# 输出学生信息和及格状态
print("学生信息：")
for student in students:
    print(student, "是否及格：", student.is_passed())


# 3. 定义 GraduateStudent 类，继承自 Student
class GraduateStudent(Student):
    def __init__(self, name, score, research_topic):
        super().__init__(name, score)
        self.research_topic = research_topic

    # 重写 __str__() 方法
    def __str__(self):
        return f"{self.name}（成绩：{self.score}，研究方向：{self.research_topic}）"


# 创建一个研究生对象并输出
graduate = GraduateStudent("Alice", 85, "大数据分析")
print("\n研究生对象：")
print(graduate)
print("是否及格：", graduate.is_passed())
```

**类与继承说明：**

1. 定义 `Student` 类，在 `__init__()` 方法中初始化姓名和成绩属性，并通过 `is_passed()` 方法判断学生是否及格。
2. 使用 `__str__()` 方法定义学生对象的字符串表示，并根据实验4中的有效记录创建多个 `Student` 对象，输出学生信息和及格状态。
3. 定义继承自 `Student` 的 `GraduateStudent` 类，增加 `research_topic` 属性，并使用 `super().__init__()` 调用父类的初始化方法。
4. 重写 `GraduateStudent` 类的 `__str__()` 方法，使研究生对象在输出时同时显示姓名、成绩和研究方向。

```python
# 实验5：直接赋值、浅拷贝与深拷贝

import copy

# 1. 直接赋值：从新建的原始对象开始
original_direct = {"name": "Alice", "scores": [85, 90]}
assigned = original_direct

assigned["scores"].append(95)

print("直接赋值：")
print("原对象：", original_direct)
print("赋值后的对象：", assigned)


# 2. 浅拷贝：重新创建原始对象
original_shallow = {"name": "Alice", "scores": [85, 90]}
shallow = copy.copy(original_shallow)

shallow["scores"].append(95)

print("\n浅拷贝：")
print("原对象：", original_shallow)
print("浅拷贝对象：", shallow)


# 3. 深拷贝：再次重新创建原始对象
original_deep = {"name": "Alice", "scores": [85, 90]}
deep = copy.deepcopy(original_deep)

deep["scores"].append(95)

print("\n深拷贝：")
print("原对象：", original_deep)
print("深拷贝对象：", deep)
```

**对象复制说明：**

1. 直接赋值时，两个变量指向同一个对象，因此通过其中一个变量修改内部列表，会影响另一个变量所指向的对象。
2. 浅拷贝使用 `copy.copy()` 创建一个新的外层对象，但不会复制内部的嵌套列表，因此修改浅拷贝对象的内部列表，原对象也会发生变化。
3. 深拷贝使用 `copy.deepcopy()` 创建新的对象，并递归复制内部的嵌套对象，因此修改深拷贝对象的内部列表，不会影响原对象。
4. 本次实验分别创建了新的原始对象进行直接赋值、浅拷贝和深拷贝操作，避免前一次操作影响后面的实验结果。
