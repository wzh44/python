```python
import numpy as np

# 创建形状为 (3, 4) 的数组
arr = np.arange(12).reshape(3, 4)

print("原始数组：")
print(arr)
print("原始数组形状：", arr.shape)

# 1. 取第 2 行
row2 = arr[1, :]
print("\n第 2 行：")
print(row2)
print("形状：", row2.shape)

# 2. 取第 1 至第 3 列
cols1to3 = arr[:, 0:3]
print("\n第 1 至第 3 列：")
print(cols1to3)
print("形状：", cols1to3.shape)

# 3. 取出所有大于 6 的元素
greater6 = arr[arr > 6]
print("\n大于 6 的元素：")
print(greater6)
print("形状：", greater6.shape)
```

### 实验结果分析

本实验使用 NumPy 数组索引和布尔筛选完成了三项操作。首先，通过 arr[1, :] 获取数组的第 2 行，结果包含 4 个元素，形状为 (4,)。其次，通过 arr[:, 0:3] 获取第 1 至第 3 列，得到一个形状为 (3, 3) 的二维数组。最后，通过 arr[arr > 6] 筛选出所有大于 6 的元素，结果为 7、8、9、10、11，形状为 (5,)。

从本实验可以看出，NumPy 支持通过行列索引提取数组中的部分数据，也可以利用条件表达式进行布尔筛选。需要注意的是，Python 的索引从 0 开始，切片的结束位置不包含在结果中。对于一维数组，shape 会显示为包含一个元素的元组，例如 (4,)。

```python
import numpy as np

# 创建包含 0 到 9 的数组
arr = np.arange(10)
print("原始数组：", arr)

# 1. 选出小于 2 或大于 7 的元素
result_or = arr[(arr < 2) | (arr > 7)]
print("\n小于 2 或大于 7 的元素：")
print(result_or)

# 使用 np.logical_or() 实现相同操作
result_or2 = arr[np.logical_or(arr < 2, arr > 7)]
print("使用 logical_or() 的结果：", result_or2)

# 2. 选出大于 3 且小于 7 的元素
result_and = arr[(arr > 3) & (arr < 7)]
print("\n大于 3 且小于 7 的元素：")
print(result_and)

# 使用 np.logical_and() 实现相同操作
result_and2 = arr[np.logical_and(arr > 3, arr < 7)]
print("使用 logical_and() 的结果：", result_and2)
```

### 实验结果分析

本实验使用 np.arange(10) 创建了包含 0 到 9 的 NumPy 数组，并通过逻辑条件筛选元素。使用 | 运算符可以实现逻辑“或”，筛选出小于 2 或大于 7 的元素，结果为 [0, 1, 8, 9]。使用 & 运算符可以实现逻辑“且”，筛选出大于 3 且小于 7 的元素，结果为 [4, 5, 6]。

此外，本实验还使用 np.logical_or() 和 np.logical_and() 完成了相同的筛选操作，结果一致。通过本实验可以看出，NumPy 支持使用布尔条件进行灵活的数据筛选。使用 | 和 & 组合多个条件时，需要用括号明确每个比较表达式，避免出现运算优先级问题。

```python
import numpy as np

# 创建两个形状为 (2, 2) 的数组
A = np.array([
    [1, 2],
    [3, 4]
])

B = np.array([
    [5, 6],
    [7, 8]
])

print("数组 A：")
print(A)

print("\n数组 B：")
print(B)

# 1. 使用 * 进行逐元素乘法
print("\n使用 * 进行逐元素乘法：")
print(A * B)

# 2. 使用 np.multiply() 进行逐元素乘法
print("\n使用 np.multiply() 进行逐元素乘法：")
print(np.multiply(A, B))

# 3. 使用 @ 进行矩阵乘法
print("\n使用 @ 进行矩阵乘法：")
print(A @ B)

# 4. 使用 np.matmul() 进行矩阵乘法
print("\n使用 np.matmul() 进行矩阵乘法：")
print(np.matmul(A, B))
```

### 实验结果分析

本实验创建了两个形状为 (2, 2) 的 NumPy 数组，并分别使用 *、np.multiply()、@ 和 np.matmul() 进行运算。

从运行结果可以看出，* 和 np.multiply() 的结果相同，它们执行的是逐元素乘法，即将两个数组中相同位置的元素分别相乘，结果为 [[5, 12], [21, 32]]。而 @ 和 np.matmul() 的结果相同，它们执行的是矩阵乘法，按照矩阵乘法规则计算行与列的乘积之和，结果为 [[19, 22], [43, 50]]。

因此，NumPy 中的 * 默认用于逐元素乘法，@ 和 np.matmul() 用于矩阵乘法。在进行数值计算时，需要根据实际需求选择合适的运算方式。

```python
import numpy as np

# 设置固定随机种子，保证结果可以复现
rng = np.random.default_rng(42)

# 生成 1000 个服从正态分布的模拟成绩
scores = rng.normal(loc=70, scale=10, size=1000)

# 将成绩限制在 0 到 100 之间
scores = np.clip(scores, 0, 100)

# 计算统计指标
sample_count = len(scores)
score_min = np.min(scores)
score_max = np.max(scores)
score_mean = np.mean(scores)
score_std = np.std(scores)
score_range = score_max - score_min

# 输出统计结果
print("样本数：", sample_count)
print("最小值：", round(score_min, 2))
print("最大值：", round(score_max, 2))
print("均值：", round(score_mean, 2))
print("标准差：", round(score_std, 2))
print("极差：", round(score_range, 2))
```

### 实验结果分析

本实验使用 NumPy 的随机数生成器生成了 1000 个服从正态分布的模拟成绩，设置分布均值为 70、标准差为 10，并使用 np.clip() 将成绩限制在 0 到 100 之间。为了保证实验结果可以复现，本实验设置了固定的随机种子 42。

通过计算可以获得成绩的样本数、最小值、最大值、均值、标准差和极差。其中，极差等于最大值减去最小值，用于反映数据的整体取值范围。由于使用了 np.clip()，所有成绩都处于 0 到 100 之间。实际样本均值和标准差不一定恰好等于设定的 70 和 10，这是因为随机生成的样本会存在一定波动，且限制成绩范围也会改变部分数据的分布。

```python
import numpy as np

# 创建整数 1 到 10
numbers = np.arange(1, 11)

# 设置固定随机种子
rng_sample = np.random.default_rng(2026)

# 1. 均匀抽样：每个整数的抽取机会相同
uniform_sample = rng_sample.choice(
    numbers,
    size=5,
    replace=False
)

print("均匀抽样结果：", uniform_sample)

# 2. 加权抽样：整数越大，权重越高
weights = numbers / numbers.sum()

weighted_sample = rng_sample.choice(
    numbers,
    size=5,
    replace=False,
    p=weights
)

print("加权抽样结果：", weighted_sample)

# 输出每个整数对应的抽样权重
print("加权抽样概率：", np.round(weights, 4))

# 检查是否存在重复元素
print("均匀抽样是否有重复：", len(set(uniform_sample)) != len(uniform_sample))
print("加权抽样是否有重复：", len(set(weighted_sample)) != len(weighted_sample))
```

### 实验结果分析

本实验从整数 1 到 10 中分别进行了均匀抽样和加权抽样，每种方式均不放回地抽取 5 个整数，并通过固定随机种子保证结果可以复现。

均匀抽样使用 np.random.Generator.choice() 方法，并将 replace 设置为 False，使每个整数在每次抽样开始时具有相同的抽取机会。加权抽样则将整数本身作为权重，除以所有整数之和得到抽样概率，因此数值越大的整数具有越高的初始抽取权重。

从运行结果可以看出，两种抽样方式都没有重复元素。均匀抽样适用于每个样本具有相同初始抽取机会的情况，而加权抽样可以根据不同样本的权重调整抽取机会。由于采用不放回抽样，每抽出一个整数后，后续抽样的概率会发生变化。

```python
import numpy as np

# 创建原始数组
arr = np.arange(1, 11)

print("原始数组：", arr)

# 1. 使用 permutation() 返回随机排列后的新数组
rng_perm = np.random.default_rng(100)

permuted_arr = rng_perm.permutation(arr)

print("\n使用 permutation() 后的新数组：")
print(permuted_arr)

print("permutation() 后的原数组：")
print(arr)

# 2. 使用 shuffle() 原地打乱数组
shuffle_arr = arr.copy()

print("\nshuffle() 操作前的数组：")
print(shuffle_arr)

rng_shuffle = np.random.default_rng(101)
rng_shuffle.shuffle(shuffle_arr)

print("shuffle() 操作后的数组：")
print(shuffle_arr)

# 检查最初的原数组是否发生变化
print("\n最初的原数组：")
print(arr)
```

### 实验结果分析

本实验分别使用 NumPy 的 permutation() 和 shuffle() 方法对数组进行随机排列，并通过固定随机种子使结果可以复现。

permutation() 会返回一个新的随机排列数组，不会修改传入的原数组。因此，可以同时保留原始数据和随机排列后的结果。shuffle() 则会直接修改传入数组中元素的顺序，不会返回一个新的排列数组。

实验中使用 copy() 创建了原数组的副本，再对副本执行 shuffle()。运行结果表明，shuffle_arr 的元素顺序发生了变化，而最初的 arr 保持不变。由此可以看出，permutation() 适合需要保留原数据的情况，而 shuffle() 适合直接修改已有数组的情况。

```python
import numpy as np
from scipy.linalg import solve

# 将方程组写成矩阵形式 A @ result = b
A = np.array([
    [3, 1],
    [1, 2]
], dtype=float)

b = np.array([9, 8], dtype=float)

# 使用 SciPy 求解线性方程组
result = solve(A, b)

x, y = result

print("方程组的解：")
print("x =", x)
print("y =", y)

# 使用 NumPy 代入验证
check_result = A @ result

print("\n代入计算结果：", check_result)
print("原方程右侧：", b)

# 判断计算结果是否与原方程右侧一致
print("验证是否通过：", np.allclose(check_result, b))
```

### 实验结果分析

本实验使用 SciPy 的 scipy.linalg.solve() 函数求解线性方程组，得到 x = 2、y = 3。随后，使用 NumPy 将求得的解代入矩阵形式的方程组，通过矩阵乘法计算验证结果。

运行结果显示，代入计算得到的结果为 [9, 8]，与方程组右侧的数值一致，验证结果为 True。这说明使用 SciPy 求得的解满足原方程组要求，也体现了 SciPy 在科学计算和线性方程组求解中的应用。


```python
import numpy as np
from scipy.optimize import root

# 定义方程 x^2 - 2 = 0
def f(x):
    return x**2 - 2

# 使用正数作为初始猜测值
positive_result = root(f, x0=1.0)

# 使用负数作为初始猜测值
negative_result = root(f, x0=-1.0)

# 输出正根的求解结果
print("正根求解是否成功：", positive_result.success)
print("正根：", positive_result.x[0])
print("正根对应的函数值：", f(positive_result.x[0]))

# 输出负根的求解结果
print("\n负根求解是否成功：", negative_result.success)
print("负根：", negative_result.x[0])
print("负根对应的函数值：", f(negative_result.x[0]))

# 检查函数值是否接近 0
print("\n正根验证是否通过：", np.isclose(f(positive_result.x[0]), 0))
print("负根验证是否通过：", np.isclose(f(negative_result.x[0]), 0))
```

### 实验结果分析

本实验使用 SciPy 的 scipy.optimize.root() 函数求解方程 x^2 - 2 = 0。为了分别寻找正根和负根，使用 1.0 和 -1.0 作为初始猜测值进行求解。

从运行结果可以看出，两次求根均成功，得到的根分别约为 1.4142 和 -1.4142。进一步将求得的根代入原方程，计算结果均接近 0，验证了求解结果的正确性。由于计算机使用浮点数进行数值计算，函数值可能存在极小的舍入误差，因此使用 np.isclose() 判断函数值是否接近 0。

```python
import numpy as np
from scipy.optimize import minimize

# 定义目标函数 f(x) = x^2 + 2x + 2
def objective(x):
    return x[0]**2 + 2 * x[0] + 2

# 使用 SciPy 求函数的最小值
result_min = minimize(objective, x0=[0.0])

# 输出数值优化结果
print("优化是否成功：", result_min.success)
print("极小点 x =", result_min.x[0])
print("函数最小值 f(x) =", result_min.fun)

# 通过配方得到解析结果
# x^2 + 2x + 2 = (x + 1)^2 + 1
x_exact = -1.0
f_exact = 1.0

print("\n解析极小点：", x_exact)
print("解析最小值：", f_exact)

# 比较数值计算结果和解析结果
print("极小点是否接近解析结果：",
      np.isclose(result_min.x[0], x_exact))
print("最小值是否接近解析结果：",
      np.isclose(result_min.fun, f_exact))
```

### 实验结果分析

本实验使用 SciPy 的 scipy.optimize.minimize() 函数对目标函数进行数值优化，获得函数的极小点和最小值，并与解析结果进行比较。

通过配方可得，目标函数可以写成 (x + 1)² + 1，因此其极小点为 x = -1，最小值为 1。运行结果显示，数值优化得到的结果与解析结果接近，且优化过程成功。通过 np.isclose() 判断，数值结果与解析结果一致。

本实验表明，SciPy 可以通过数值优化方法求解函数的极小值，并可以使用解析结果对数值计算结果进行验证。


```python
import numpy as np
from scipy.stats import ttest_ind

# 设置固定随机种子
rng_t = np.random.default_rng(2026)

# 生成第一组数据：均值为 0，标准差为 1
group1 = rng_t.normal(loc=0, scale=1, size=100)

# 生成第二组数据：均值为 1，标准差为 1
group2 = rng_t.normal(loc=1, scale=1, size=100)

# 执行独立样本 t 检验
test_result = ttest_ind(group1, group2)

# 输出样本信息
print("第一组样本数：", len(group1))
print("第二组样本数：", len(group2))
print("第一组样本均值：", round(np.mean(group1), 4))
print("第二组样本均值：", round(np.mean(group2), 4))

# 输出 t 检验结果
print("\nt 统计量：", round(test_result.statistic, 4))
print("p 值：", test_result.pvalue)

# 在 0.05 显著性水平下进行判断
alpha = 0.05

if test_result.pvalue < alpha:
    print("检验结论：拒绝原假设，两组总体均值存在显著差异。")
else:
    print("检验结论：不能拒绝原假设，证据不足以表明两组总体均值存在显著差异。")
```

### 实验结果分析

本实验使用 NumPy 的随机数生成器生成了两组独立的模拟样本，每组包含 100 个观测值。第一组数据来自均值为 0、标准差为 1 的正态分布，第二组数据来自均值为 1、标准差为 1 的正态分布。实验设置了固定随机种子，以保证结果可以复现。

随后，使用 SciPy 的 scipy.stats.ttest_ind() 函数进行独立样本 t 检验，并计算 t 统计量和 p 值。在显著性水平 0.05 下，当 p 值小于 0.05 时，拒绝两组总体均值相等的原假设；否则不能拒绝原假设。通过该实验，可以了解如何使用 SciPy 对两组独立样本的总体均值进行统计检验。

```python
# 选择三科成绩
subject_scores = students_df[["Math", "English", "Python"]]

# 1. 使用 apply() 计算每名学生的三科最高分
students_df["最高分"] = subject_scores.apply(
    lambda row: row.max(),
    axis=1
)

print("每名学生的三科最高分：")
print(students_df[["Math", "English", "Python", "最高分"]])

# 2. 使用 agg() 一次计算每科的最小值、均值和最大值
print("\n每科成绩的最小值、均值和最大值：")
print(subject_scores.agg(["min", "mean", "max"]))

# 3. 使用 rank() 对 Python 成绩排名
students_df["Python排名"] = students_df["Python"].rank(
    ascending=False,
    method="min"
)

print("\n学生 Python 成绩排名：")
print(students_df[["Python", "Python排名"]].sort_values("Python排名"))
```

最后，使用 rank() 对学生的 Python 成绩进行排名，并设置 ascending=False，使成绩越高排名越靠前。运行结果显示，Alice 的 Python 成绩为 95 分，排名第一；David 的 Python 成绩为 70 分，排名第六。

通过本实验可以看出，apply() 可以对行或列应用自定义计算，agg() 可以一次执行多种统计操作，而 rank() 可以根据数值大小生成排名。

通过本实验可以看出，filter() 可以根据每个分组整体满足的条件，决定是否保留该组的所有记录。它与直接筛选单个学生不同，更适合根据组内统计指标进行分组筛选。

```python
# 选择三科成绩
subject_scores = students_df[["Math", "English", "Python"]]

# 1. 计算协方差矩阵
cov_matrix = subject_scores.cov()

print("三科成绩的协方差矩阵：")
print(cov_matrix)

# 2. 计算皮尔森相关系数矩阵
corr_matrix = subject_scores.corr()

print("\n三科成绩的皮尔森相关系数矩阵：")
print(corr_matrix)

# 3. 选择数学和 Python 两科进行分析
cov_math_python = cov_matrix.loc["Math", "Python"]
corr_math_python = corr_matrix.loc["Math", "Python"]

print("\n数学与 Python 成绩的协方差：", round(cov_math_python, 4))
print("数学与 Python 成绩的相关系数：", round(corr_math_python, 4))
```

### 实验结果分析

本实验使用 pandas 的 cov() 和 corr() 方法计算了数学、英语和 Python 三科成绩的协方差矩阵与皮尔森相关系数矩阵。

协方差矩阵反映不同科目成绩之间的共同变化趋势。以数学和 Python 成绩为例，两者的协方差约为 74.9333，结果为正，说明在这组数据中，两科成绩总体上具有同向变化的趋势。

皮尔森相关系数用于衡量两个变量之间线性相关关系的方向和强度，其取值范围为 -1 到 1。数学与 Python 成绩的相关系数约为 0.8848，接近 1，说明两科成绩在这组样本中具有较强的正线性相关关系。

需要注意，相关性不代表因果关系，不能仅凭相关系数判断某一科成绩的变化是否导致另一科成绩发生变化。
