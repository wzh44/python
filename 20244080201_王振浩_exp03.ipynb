```python
import hashlib
import os
import platform
import uuid
import sys

# 填写自己的学号
STUDENT_ID = 20244080201

# 获取当前 Conda 环境
CONDA_ENV = os.environ.get("CONDA_DEFAULT_ENV")

print(f"[Conda 环境] {CONDA_ENV}")

# 生成机器指纹
MACHINE_FP = hashlib.sha256(
    f"{platform.node()}|{platform.machine()}|{platform.system()}|{uuid.getnode()}".encode()
).hexdigest()[:16]

# 生成运行编号
RUN_ID = f"{STUDENT_ID}-{MACHINE_FP}"

print(f"[学号] {STUDENT_ID}")
print(f"[机器指纹] {MACHINE_FP}")
print(f"[RUN_ID] {RUN_ID}")
print(f"[Python] {sys.version.split()[0]}")
```

```python
import re

text = "cat on mat; later: cats and dogs"

# 匹配数字
numbers = re.findall(r'\d+', text)

# 匹配空白字符
spaces = re.findall(r'\s+', text)

# 匹配独立的 cat 单词
cats = re.findall(r'\bcat\b', text)

print("原始文本：", text)
print("数字匹配结果：", numbers)
print("空白字符匹配结果：", spaces)
print("独立 cat 匹配结果：", cats)
```

```python
# 使用 | 匹配 cat 或 dog
text = "cat dog bird CAT"

result = re.findall(r'cat|dog', text)

print("原始文本：", text)
print("匹配结果：", result)
text = "aaaa"

result1 = re.findall(r"a|aa|aaa", text)
result2 = re.findall(r"aaa|aa|a", text)

print("正则表达式 a|aa|aaa：", result1)
print("正则表达式 aaa|aa|a：", result2)
```

```python
text = "ct cat caat caaat"

patterns = [
    r'ca*t',
    r'ca+t',
    r'ca?t',
    r'ca{2}t',
    r'ca{1,2}t'
]

for pattern in patterns:
    result = re.findall(pattern, text)
    print(f"{pattern} 的匹配结果：{result}")
```

```python
import re

text = "联系：alice@example.com；备选：bob@test.org"

# 使用捕获组提取邮箱用户名和域名
pattern = r'(\w+)@([\w.]+)'

results = re.findall(pattern, text)

print("原始文本：", text)
print("邮箱提取结果：")

for username, domain in results:
    print("用户名：", username, "域名：", domain)
```

```python
text = "cat cat, dog cat, bird bird"

# 使用后向引用查找连续重复的单词
pattern = r'\b(\w+)\s+\1\b'

result = re.findall(pattern, text)

print("原始文本：", text)
print("连续重复的单词：", result)
```

```python
text = "2024-05-01\n2024/05/02\n2024.05.03"

# 使用正则表达式提取年月日，并转换顺序
pattern = r'(\d{4})[-/.](\d{2})[-/.](\d{2})'

replacement = r'\2-\3-\1'

new_text = re.sub(pattern, replacement, text)

print("转换前：")
print(text)

print("\n转换后：")
print(new_text)
```

```python
text = "联系我：123-456-7890 或 987.654.3210"

# 匹配电话号码
pattern = r'(\d{3})[-.](\d{3})[-.](\d{4})'

# 统一电话号码格式
replacement = r'(\1) \2-\3'

new_text = re.sub(pattern, replacement, text)

print("转换前：", text)
print("转换后：", new_text)
```

```python
from bs4 import BeautifulSoup

print("BeautifulSoup 导入成功！")
```

```python
html_doc = """
<html><head><title>实验课程</title></head><body>
<h1>数据分析</h1>
<p class="course">第六讲：正则表达式</p>
<p class="course">第七讲：结构化数据</p>
<a href="https://example.org/lecture06">讲义六</a>
<a href="https://example.org/lecture07">讲义七</a>
</body></html>
"""

# 构建 HTML 解析树
soup = BeautifulSoup(html_doc, "html.parser")

# 获取页面标题
print("页面标题：", soup.title.get_text())

# 获取 h1 标签名和文本
print("h1 标签名：", soup.h1.name)
print("h1 文本：", soup.h1.get_text())
```

```python
# 找出所有 class="course" 的段落
courses = soup.find_all("p", class_="course")

print("课程段落：")
for p in courses:
    print(p.get_text())

# 找出所有链接
links = soup.find_all("a")

print("\n所有链接：")
for a in links:
    print("链接文字：", a.get_text())
    print("href：", a.get("href"))
```

```python
# 提取整个 HTML 文档的可见文本
print("整个文档的可见文本：")
print(soup.get_text(separator="\n", strip=True))

# 提取单个课程段落的文本
print("\n单个段落文本：")
print(courses[0].get_text())

# 提取单个链接的文本
print("\n单个链接文本：")
print(links[0].get_text())
```
