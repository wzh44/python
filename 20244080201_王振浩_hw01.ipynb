```python
# HW01 第二部分：欧几里得算法

def gcd(m, n):
    # 将两个数转换为非负数
    m = abs(m)
    n = abs(n)

    # 使用欧几里得算法计算最大公约数
    while n != 0:
        m, n = n, m % n

    return m


# 测试老师要求的三组数字
print("GCD(2018, 2019) =", gcd(2018, 2019))
print("GCD(1200, 300) =", gcd(1200, 300))
print("GCD(5040, 60) =", gcd(5040, 60))

# 测试负数情况
print("GCD(-12, 8) =", gcd(-12, 8))
print("GCD(12, -8) =", gcd(12, -8))
print("GCD(-12, -8) =", gcd(-12, -8))
```

**欧几里得算法说明：**

1. 使用 `def gcd(m, n)` 定义函数，接收两个整数作为参数。
2. 使用 `abs()` 将两个参数转换为非负数，保证计算结果符合通常的最大公约数定义。
3. 使用 `while` 循环和取余运算实现欧几里得算法。当 `n` 不为0时，将 `m` 更新为 `n`，将 `n` 更新为 `m % n`。当 `n` 等于0时，返回 `m` 作为最大公约数。
4. 三组测试数据的最大公约数分别为1、300和60。
5. 当输入一个或两个负数时，函数会先使用 `abs()` 将负数转换为对应的非负数，因此仍然能够得到非负的最大公约数。这种处理方式是合理的，因为最大公约数通常定义为正的最大公约数；当两个输入都为0时，按照本函数的实现会返回0。

```python
# HW01 第三部分第1题：使用极限公式近似计算 e

def euler_limit(n):
    return float((1 + 1 / n) ** n)


# 测试不同的 n
for n in [1, 2, 5, 10, 100, 1000]:
    print(f"n = {n}, e 的近似值 = {euler_limit(n)}")
```

**使用极限公式近似计算 e 的说明：**

1. 使用 `def euler_limit(n)` 定义函数，接收一个整数 n。
2. 根据公式 `e ≈ (1 + 1/n) ** n` 计算自然常数 e 的近似值，并使用 `float()` 返回浮点数结果。
3. 使用不同的 n 进行测试，可以发现随着 n 增大，计算结果逐渐接近自然常数 e，说明该极限公式可以用于近似计算 e。

```python
# HW01 第三部分第2题：使用级数近似计算 e

def euler_infinite_sum(n):
    total = 0.0
    factorial = 1

    for k in range(n):
        if k > 0:
            factorial *= k
        total += 1 / factorial

    return float(total)


# 测试不同的 n
for n in [0, 1, 2, 4, 10]:
    print(f"n = {n}, e 的近似值 = {euler_infinite_sum(n)}")
```

**使用级数近似计算 e 的说明：**

1. 使用 `def euler_infinite_sum(n)` 定义函数，接收一个非负整数 n，返回前 n 项之和。
2. 公式中的每一项为 `1 / k!`。程序通过循环计算阶乘，并将每一项累加到 `total` 中。
3. 当 n 为0时，循环不执行，返回0；当 n 为1时，只累加第一项，返回1。
4. 当 n 为4时，计算结果约为2.6667；当 n 为10时，计算结果约为2.71828。随着累加项数增加，近似值逐渐接近自然常数 e。

```python
# HW01 第三部分第3题：根据误差要求近似计算 e

import math

def euler_approx(epsilon):
    total = 0.0
    factorial = 1
    k = 0

    while True:
        if k > 0:
            factorial *= k

        total += 1 / factorial

        # 判断近似值与真实值的误差是否满足要求
        if abs(total - math.exp(1)) < epsilon:
            return float(total)

        k += 1


# 测试不同的误差要求
for epsilon in [0.1, 0.01, 0.001, 0.0001]:
    result = euler_approx(epsilon)
    error = abs(result - math.exp(1))

    print(f"允许误差：{epsilon}")
    print(f"e 的近似值：{result}")
    print(f"实际误差：{error}")
    print()
```

**根据误差要求近似计算 e 的说明：**

1. 使用 `def euler_approx(epsilon)` 定义函数，参数 `epsilon` 表示允许的误差。
2. 使用无穷级数 `e = 1/0! + 1/1! + 1/2! + ...` 逐项计算自然常数 e 的近似值。
3. 使用 `while` 循环不断累加各项，并使用 `math.exp(1)` 获取自然常数 e 的参考值。
4. 每次累加后，使用 `abs(total - math.exp(1))` 计算近似值与参考值之间的绝对误差。当误差小于 `epsilon` 时，停止循环并返回近似值。
5. 通过测试不同的允许误差，可以发现误差要求越小，通常需要累加更多的项才能满足要求。

```python
# HW01 第三部分第4题：输出 e 的近似值表格

def print_euler_sum_table(n):
    for k in range(1, n + 1):
        print(f"n = {k}, e 的近似值 = {euler_infinite_sum(k)}")


# 测试函数，输出前10项的计算结果
print_euler_sum_table(10)
```

**输出 e 的近似值表格说明：**

1. 使用 `def print_euler_sum_table(n)` 定义函数，参数 `n` 表示需要输出的项数。
2. 使用 `for` 循环和 `range(1, n + 1)`，依次取出从1到n的整数。
3. 每次循环调用 `euler_infinite_sum(k)` 计算前k项的和，并使用 `print()` 将结果单独输出一行。
4. 通过输出前10项的计算结果可以发现，随着累加项数增加，计算得到的近似值逐渐接近自然常数 e。

```python
# HW01 第四部分第1题：判断一个数是不是2的幂

def is_power_of_2(n):
    while n % 2 == 0:
        n = n // 2

    return n == 1


# 测试不同的正整数
for n in [1, 2, 4, 8, 16, 18, 32, 64]:
    print(f"{n} 是不是2的幂：{is_power_of_2(n)}")
```

**判断一个数是否为2的幂的说明：**

1. 使用 `def is_power_of_2(n)` 定义函数，接收一个正整数 n，并返回布尔值。
2. 使用 `while` 循环判断 n 是否能被2整除。如果余数为0，就使用整除运算将 n 除以2。
3. 当 n 不能继续被2整除时，检查 n 是否等于1。如果等于1，说明原数是2的幂，返回 `True`；否则返回 `False`。
4. 本题使用整除运算 `//` 和取余运算 `%` 完成判断，没有使用 `math.sqrt()`。
5. 测试结果表明，1、2、4、8、16、32和64是2的幂，而18不是2的幂。

```python
# HW01 第四部分第2题：判断一个数是否为另一个数的幂

def is_power(b, n):
    # 当底数为1时，只有1是它的幂
    if b == 1:
        return n == 1

    # 不断判断是否能被底数整除
    while n % b == 0:
        n = n // b

    return n == 1


# 测试不同的数字
print("8 是不是 2 的幂：", is_power(2, 8))
print("27 是不是 3 的幂：", is_power(3, 27))
print("10 是不是 3 的幂：", is_power(3, 10))
print("125 是不是 5 的幂：", is_power(5, 125))
print("1 是不是 7 的幂：", is_power(7, 1))
print("8 是不是 1 的幂：", is_power(1, 8))
```

**判断一个数是否为另一个数的幂的说明：**

1. 使用 `def is_power(b, n)` 定义函数，接收底数 b 和待判断的正整数 n，并返回布尔值。
2. 当底数 b 为1时，只有 n 为1才返回 `True`，因为1的任意非负整数次幂都等于1。
3. 当底数大于1时，使用 `while` 循环不断判断 n 是否能被 b 整除。如果可以整除，就使用 `//` 将 n 除以 b。
4. 当 n 不能继续被 b 整除时，如果最终 n 等于1，说明原数是 b 的幂，返回 `True`；否则返回 `False`。
5. 测试结果表明，8是2的幂，27是3的幂，10不是3的幂，125是5的幂，1是7的幂，而8不是1的幂。
