```python
def is_palindrome(text):
    cleaned = text.replace(" ", "").lower()
    return cleaned == cleaned[::-1]


# 测试函数
print(is_palindrome("level"))
print(is_palindrome("T A C O cat"))
print(is_palindrome("hello"))
print(is_palindrome("12321"))
print(is_palindrome("python"))
```

```python
def is_abecedarian(text):
    cleaned = text.replace(" ", "").lower()
    
    for i in range(len(cleaned) - 1):
        if cleaned[i] > cleaned[i + 1]:
            return False
    
    return True


# 测试函数
print(is_abecedarian("adder"))
print(is_abecedarian("beet"))
print(is_abecedarian("dog"))
print(is_abecedarian("cat"))
print(is_abecedarian("abcd efgh xyz"))
```

```python
def remove_vowels(text):
    vowels = "aeiou"
    return "".join(char for char in text if char.lower() not in vowels)


# 测试函数
print(remove_vowels("cat"))
print(remove_vowels("audio"))
print(remove_vowels("hello"))
print(remove_vowels("aeiou"))
print(remove_vowels("python"))
print(remove_vowels("EDUCATION"))
```

```python
def list_reverse(lst):
    if not isinstance(lst, list):
        raise TypeError("输入参数必须是列表")

    return lst[::-1]


# 测试函数
print(list_reverse([1, 2, 3]))
print(list_reverse(["a", "b", "c"]))
print(list_reverse([5]))
print(list_reverse([]))
```

```python
def is_sorted(seq):
    for i in range(len(seq) - 1):
        if seq[i] > seq[i + 1]:
            return False

    return True


# 测试函数
print(is_sorted([1, 2, 3, 4]))
print(is_sorted([1, 1, 2, 3]))
print(is_sorted([3, 2, 1]))
print(is_sorted([1, 3, 2, 4]))
print(is_sorted([]))
print(is_sorted([5]))
```

```python
def vec_scalar_mult(t, s):
    if not isinstance(t, tuple):
        raise TypeError("向量必须是元组")

    if not all(type(x) in (int, float) for x in t):
        raise TypeError("向量元素必须是整数或浮点数")

    if type(s) not in (int, float):
        raise TypeError("标量必须是整数或浮点数")

    return tuple(x * s for x in t)


# 测试函数
print(vec_scalar_mult((1, 2, 3), 2))
print(vec_scalar_mult((2, 4), 0.5))
print(vec_scalar_mult((-1, 3), 2))
print(vec_scalar_mult((), 5))
```

```python
def vec_inner_product(t1, t2):
    if not isinstance(t1, tuple) or not isinstance(t2, tuple):
        raise TypeError("两个向量都必须是元组")

    if len(t1) != len(t2):
        raise ValueError("两个向量的长度必须一致")

    if not all(type(x) in (int, float) for x in t1 + t2):
        raise TypeError("向量元素必须是整数或浮点数")

    return float(sum(x * y for x, y in zip(t1, t2)))


# 测试函数
print(vec_inner_product((1, 2, 3), (4, 5, 6)))
print(vec_inner_product((2, 3), (4, 5)))
print(vec_inner_product((1.5, 2.0), (2, 3)))
print(vec_inner_product((), ()))
```

```python
def check_valid_mx(mx):
    # 检查外层是否为元组
    if not isinstance(mx, tuple):
        return False

    # 检查每一行是否都是元组
    if not all(isinstance(row, tuple) for row in mx):
        return False

    # 检查每一行的长度是否相同
    if len(mx) > 0:
        row_length = len(mx[0])
        for row in mx:
            if len(row) != row_length:
                return False

    # 检查所有元素是否为整数或浮点数
    for row in mx:
        for x in row:
            if type(x) not in (int, float):
                return False

    return True


# 测试函数
print(check_valid_mx(((1, 2), (3, 4))))
print(check_valid_mx(((1, 2, 3), (4, 5, 6))))
print(check_valid_mx(((1, 2), (3,))))
print(check_valid_mx(((1, "a"), (3, 4))))
print(check_valid_mx([[1, 2], [3, 4]]))
```

```python
def mx_vec_mult(mx, v):
    # 检查矩阵是否合法
    if not check_valid_mx(mx):
        raise ValueError("矩阵格式不合法")

    # 检查向量是否为元组
    if not isinstance(v, tuple):
        raise TypeError("向量必须是元组")

    # 检查向量元素类型
    if not all(type(x) in (int, float) for x in v):
        raise TypeError("向量元素必须是整数或浮点数")

    # 检查向量长度是否等于矩阵列数
    if len(mx) > 0 and len(v) != len(mx[0]):
        raise ValueError("向量长度必须等于矩阵的列数")

    # 逐行计算矩阵与向量的乘积
    result = []
    for row in mx:
        value = sum(x * y for x, y in zip(row, v))
        result.append(value)

    return tuple(result)


# 测试函数
mx1 = (
    (1, 2, 3),
    (4, 5, 6)
)

v1 = (1, 2, 3)

print(mx_vec_mult(mx1, v1))

mx2 = (
    (1, 2),
    (3, 4),
    (5, 6)
)

v2 = (2, 3)

print(mx_vec_mult(mx2, v2))
```

```python
class Vector:
    def __init__(self, dimension, elements=None):
        # 检查维度
        if type(dimension) is not int:
            raise TypeError("维度必须是整数")

        if dimension < 0:
            raise ValueError("维度不能为负数")

        # 默认创建零向量
        if elements is None:
            elements = (0,) * dimension

        # 检查元素容器
        if not isinstance(elements, (list, tuple)):
            raise TypeError("向量元素必须是列表或元组")

        # 检查元素类型
        if not all(type(x) in (int, float) for x in elements):
            raise TypeError("向量元素必须是整数或浮点数")

        # 检查维度和元素数量
        if len(elements) != dimension:
            raise ValueError("维度与元素数量不一致")

        self.dimension = dimension
        self.elements = tuple(elements)

    # 检查参与运算的对象
    def _check_other(self, other):
        if not isinstance(other, Vector):
            raise TypeError("参数必须是 Vector 对象")

        if self.dimension != other.dimension:
            raise ValueError("两个向量的维度必须一致")

    # 等于
    def __eq__(self, other):
        self._check_other(other)
        return self.elements == other.elements

    # 小于
    def __lt__(self, other):
        self._check_other(other)
        return self.elements < other.elements

    # 小于或等于
    def __le__(self, other):
        self._check_other(other)
        return self.elements <= other.elements

    # 大于
    def __gt__(self, other):
        self._check_other(other)
        return self.elements > other.elements

    # 大于或等于
    def __ge__(self, other):
        self._check_other(other)
        return self.elements >= other.elements

    # 计算两个向量的内积
    def dot(self, other):
        self._check_other(other)
        return float(
            sum(x * y for x, y in zip(self.elements, other.elements))
        )

    # 实现乘法运算符
    def __mul__(self, other):
        # 两个向量相乘：对应位置的元素分别相乘
        if isinstance(other, Vector):
            self._check_other(other)

            result = tuple(
                x * y for x, y in zip(self.elements, other.elements)
            )

            return Vector(self.dimension, result)

        # 向量乘以标量
        if type(other) in (int, float):
            result = tuple(x * other for x in self.elements)
            return Vector(self.dimension, result)

        raise TypeError("只能与 Vector 对象或数字相乘")

    # 实现标量左乘
    def __rmul__(self, other):
        return self.__mul__(other)
```

```python
# 创建一个三维向量
v1 = Vector(3, (1, 2, 3))
print(v1.dimension)
print(v1.elements)

# 只提供维度，默认创建零向量
v2 = Vector(3)
print(v2.dimension)
print(v2.elements)

# 使用列表作为向量元素
v3 = Vector(2, [1.5, 2.5])
print(v3.elements)

# 创建零维向量
v4 = Vector(0)
print(v4.elements)
```

```python
# 创建三个二维向量
v1 = Vector(2, (2, 4))
v2 = Vector(2, (2, 5))
v3 = Vector(2, (2, 4))

# 测试五种比较运算符
print(v1 == v3)
print(v1 == v2)
print(v1 < v2)
print(v1 <= v3)
print(v2 > v1)
print(v2 >= v1)

# 测试维度不同的情况
try:
    v1 < Vector(3, (1, 2, 3))
except ValueError as e:
    print(type(e).__name__, e)
```

```python
# 测试两个三维向量的内积
v1 = Vector(3, (1, 2, 3))
v2 = Vector(3, (4, 5, 6))

print(v1.dot(v2))

# 测试两个二维向量的内积
v3 = Vector(2, (2, 3))
v4 = Vector(2, (4, 5))

print(v3.dot(v4))

# 测试零向量
v5 = Vector(3)
print(v5.dot(v1))

# 测试维度不一致的情况
try:
    v1.dot(Vector(2, (1, 2)))
except ValueError as e:
    print(type(e).__name__, e)

# 测试参数类型错误的情况
try:
    v1.dot((1, 2, 3))
except TypeError as e:
    print(type(e).__name__, e)
```
